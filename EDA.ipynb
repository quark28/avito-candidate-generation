```python
import re

def rating_thr(s):
    m = re.search(r"Рейтинг[^0-9]*(\d(?:[.,]\d)?)", s)
    return float(m.group(1).replace(",", ".")) if m else None

sub = train[train["search_infm_params_text"].str.contains("Рейтинг", na=False)].copy()
sub["thr"] = sub["search_infm_params_text"].apply(rating_thr)

print(sub["search_infm_params_text"].str.extract(r"(Рейтинг[^А-ЯA-Z]*)")[0].value_counts(), "\n")
print("не распарсилось:", sub["thr"].isna().sum())

sub["ok"] = sub["item_rating"] >= sub["thr"]
print(sub.groupby("thr").agg(
    поисков=("ok", "size"),
    соблюдают=("ok", "mean"),
    рейтинг_nan=("item_rating", lambda x: x.isna().mean()),
))
```

```python
Q_COLS = ["search_query", "search_location_id", "search_is_delivery_search",
          "search_infm_params_text", "search_category"]

g = train.groupby(Q_COLS, dropna=False)["item_id"].agg(rows="size", uniq="nunique")

multi = g[g["rows"] > 1].copy()
multi["repeat"] = multi["rows"] - multi["uniq"]

print("запросов с >1 выбором:", len(multi))
print("из них хоть раз выбрали одно и то же объявление дважды:", (multi["repeat"] > 0).mean())
print("доля повторных выборов среди всех выборов в таких запросах:",
      multi["repeat"].sum() / multi["rows"].sum())

multi["bucket"] = pd.cut(multi["rows"], [1, 2, 5, 10, 50, 10_000])
print(multi.groupby("bucket", observed=True).agg(
    запросов=("rows", "size"),
    доля_повторов=("repeat", lambda r: r.sum() / multi.loc[r.index, "rows"].sum()),
))
```

```python
import re

KEYS = r"(?:Тип услуги|Место оказания услуг|Рейтинг пользователя|$)"
sub = train[train["search_infm_params_text"].str.contains("Вид услуги", na=False)].copy()

sub["val"] = (sub["search_infm_params_text"]
              .str.extract(rf"Вид услуги\s*(.*?)\s*{KEYS}")[0]
              .fillna("").str.strip())

empty = sub["val"] == ""
print("поисков с 'Вид услуги':", len(sub))
print("из них пустой фильтр (фильтра нет):", empty.mean())

real = sub[~empty].copy()
real["ok"] = [v in (p or "") for v, p in zip(real["val"], real["item_infm_params_text"])]
print("с реальным фильтром:", len(real), "| соблюдают:", real["ok"].mean())

print(real.loc[~real["ok"], "val"].value_counts().head(15))
pd.set_option("display.max_colwidth", 200)
print(real.loc[~real["ok"], ["search_infm_params_text", "item_infm_params_text"]].head(10))
```
